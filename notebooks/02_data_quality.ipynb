# 02 - Data Cleaning

Inspect missing values, duplicates and data inconsistencies.

## Step 1: Load the Dataset

This cell imports the required libraries, loads the customer support dataset, and displays its dimensions.

```python
import pandas as pd
import numpy as np

# Load dataset
df = pd.read_csv("../data/raw/customer_support_tickets.csv")

print("Dataset Loaded Successfully!")
print("-" * 40)

print("Total Rows:", df.shape[0])
print("Total Columns:", df.shape[1])
```

**Summary:** The dataset is loaded into a Pandas DataFrame. The row and column counts confirm that the file was read successfully.

## Step 2: Understand the Dataset

This cell previews sample records, checks column data types and non-null counts, and generates a statistical summary.

```python
# First five records
display(df.head())

# Dataset information
df.info()

# Statistical summary
display(df.describe(include="all"))
```

**Summary:** These checks provide an initial understanding of the dataset's structure, values, and distribution.

## Step 3: Missing Value Analysis

This cell counts missing values in each column and calculates their percentage to identify incomplete fields.

```python
# Missing value analysis

missing_count = df.isnull().sum()

missing_percentage = (missing_count / len(df)) * 100

missing_data = pd.DataFrame({
    'Missing Values': missing_count,
    'Missing Percentage': missing_percentage.round(2)
})

missing_data = missing_data[missing_data['Missing Values'] > 0]

print("MISSING VALUE ANALYSIS")
print("-" * 40)

display(missing_data.sort_values(
    by='Missing Values',
    ascending=False
))
```

**Summary:** The analysis highlights columns with missing information. Missing values are retained rather than replaced with arbitrary values.

# DUPLICATES DETECTION

## Step 4: Duplicate Detection

This cell checks for duplicate rows and repeated Ticket IDs, and displays duplicates if any are found.

```python
# 1. Check duplicate rows
duplicate_rows = df.duplicated().sum()

print("Duplicate Rows:", duplicate_rows)

# 2. Check duplicate Ticket IDs
duplicate_ids = df['Ticket ID'].duplicated().sum()

print("Duplicate Ticket IDs:", duplicate_ids)

# 3. Display duplicates if available
if duplicate_rows > 0:
    display(df[df.duplicated(keep=False)])
else:
    print("No duplicate records found.")

if duplicate_ids > 0:
    display(df[df['Ticket ID'].duplicated(keep=False)])
else:
    print("All Ticket IDs are unique.")
```

**Summary:** This helps verify record uniqueness and prevents repeated tickets from affecting downstream analysis.

# INVALID VALUE DETECTION

## Step 5: Invalid Value Detection

This cell checks customer ages, satisfaction ratings, and numerical fields for values outside the expected ranges.

```python
# 1. Customer Age validation

invalid_age = df[
    (df['Customer Age'] < 0) |
    (df['Customer Age'] > 120)
]

print("Invalid Customer Ages:", len(invalid_age))


# 2. Customer Satisfaction validation

invalid_rating = df[
    (df['Customer Satisfaction Rating'] < 1) |
    (df['Customer Satisfaction Rating'] > 5)
]

print("Invalid Satisfaction Ratings:", len(invalid_rating))


# 3. Check negative values in numerical columns

print("\nNegative Value Check:")

for col in ['Customer Age', 'Customer Satisfaction Rating']:
    print(col, ":", (df[col] < 0).sum())
```

**Summary:** The checks help identify potentially invalid records. The satisfaction rating range should be confirmed against the dataset's official definition.

# INCONSISTENT VALUES DETECTION

## Step 6: Inconsistent Value Detection

This cell reviews the unique values in key categorical columns and checks text fields for leading or trailing spaces.

```python
# Categorical consistency check

columns = [
    'Ticket Status',
    'Ticket Priority',
    'Ticket Channel',
    'Ticket Type',
    'Customer Gender'
]

for col in columns:

    print("\n", col)
    print("-" * 30)

    print("Unique Values:", df[col].nunique())
    print(df[col].value_counts(dropna=False))

    print("\nWHITESPACE CHECK")

for col in df.select_dtypes(include='object').columns:

    count = df[col].dropna().apply(
        lambda x: x != x.strip()
    ).sum()

    if count > 0:
        print(col, ":", count, "inconsistent values")

print("Whitespace check completed.")
```

**Summary:** This helps identify inconsistent category labels and formatting issues that could affect grouping and analysis.

# DATA TYPE VALIDATION

## Step 7: Data Type Validation

This cell inspects existing data types and converts date and numerical columns into suitable formats.

```python
# Check current data types

print("CURRENT DATA TYPES")
print("-" * 35)

print(df.dtypes)

# Convert date columns
date_columns = [
    'Date of Purchase',
    'First Response Time',
    'Time to Resolution'
]

for col in date_columns:
    df[col] = pd.to_datetime(df[col], errors='coerce')

# Convert numerical columns
df['Customer Age'] = pd.to_numeric(
    df['Customer Age'], errors='coerce'
)

df['Customer Satisfaction Rating'] = pd.to_numeric(
    df['Customer Satisfaction Rating'], errors='coerce'
)

print("\nUPDATED DATA TYPES")
print("-" * 35)

print(df.dtypes)
```

**Summary:** Correct data types support reliable filtering, calculations, and time-based analysis. Values that cannot be converted become missing and should be reviewed.

# USABLE RECORDS

## Step 8: Usable Records Assessment

This cell checks whether essential ticket fields are present and counts complete and incomplete records.

```python
# Identify essential columns for analysis
required_columns = [
    'Ticket ID',
    'Ticket Status',
    'Ticket Priority',
    'Ticket Type',
    'Ticket Channel'
]

# Check missing values in essential columns
print("MISSING VALUES IN ESSENTIAL COLUMNS")
print("-" * 40)
print(df[required_columns].isnull().sum())

# Count usable and incomplete records
usable_records = df[required_columns].notnull().all(axis=1)

print("\nUSABLE RECORDS:", usable_records.sum())
print("INCOMPLETE RECORDS:", (~usable_records).sum())
print("TOTAL RECORDS:", len(df))
```

**Summary:** Records are assessed using Ticket ID, Ticket Status, Ticket Priority, Ticket Type, and Ticket Channel. No records are deleted in this step.

# DATA CLEANING

## Step 9: Data Cleaning

This cell creates a separate copy of the dataset, removes extra spaces from text values, and checks the cleaned result.

```python
# Create a separate cleaned dataset
cleaned_df = df.copy()

# Remove unnecessary spaces from text columns
text_columns = cleaned_df.select_dtypes(include='object').columns

for col in text_columns:
    cleaned_df[col] = cleaned_df[col].str.strip()

# Check the cleaned dataset
print("Original dataset shape:", df.shape)
print("Cleaned dataset shape:", cleaned_df.shape)

print("\nRemaining duplicate rows:", cleaned_df.duplicated().sum())
print("Total missing values:", cleaned_df.isnull().sum().sum())

print("\nCleaned dataset preview:")
print(cleaned_df.head())
```

**Summary:** The original DataFrame is preserved, and no rows or missing values are removed or filled during this conservative cleaning step.
